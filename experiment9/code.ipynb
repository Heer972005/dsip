```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# Read image
img = cv2.imread("image.png")
img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

# Split channels
r, g, b = cv2.split(img)

# Create colored channel images
red = cv2.merge([r, np.zeros_like(r), np.zeros_like(r)])
green = cv2.merge([np.zeros_like(g), g, np.zeros_like(g)])
blue = cv2.merge([np.zeros_like(b), np.zeros_like(b), b])

# Display
plt.figure(figsize=(16,5))

plt.subplot(1,4,1)
plt.imshow(img)
plt.title("Original Image")
plt.axis("off")

plt.subplot(1,4,2)
plt.imshow(red)
plt.title("Red Channel")
plt.axis("off")

plt.subplot(1,4,3)
plt.imshow(green)
plt.title("Green Channel")
plt.axis("off")

plt.subplot(1,4,4)
plt.imshow(blue)
plt.title("Blue Channel")
plt.axis("off")

plt.show()
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# Read image
img = cv2.imread("image.png")
rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
hsv = cv2.cvtColor(img, cv2.COLOR_BGR2HSV)

# HSV ranges
# Red has two ranges because it wraps around the HSV hue circle
lower_red1 = np.array([0, 80, 80])
upper_red1 = np.array([10, 255, 255])
lower_red2 = np.array([170, 80, 80])
upper_red2 = np.array([180, 255, 255])

lower_green = np.array([35, 80, 80])
upper_green = np.array([85, 255, 255])

lower_blue = np.array([90, 80, 80])
upper_blue = np.array([130, 255, 255])

# Masks
red_mask = cv2.inRange(hsv, lower_red1, upper_red1) | cv2.inRange(hsv, lower_red2, upper_red2)
green_mask = cv2.inRange(hsv, lower_green, upper_green)
blue_mask = cv2.inRange(hsv, lower_blue, upper_blue)

# Extract only those colors
red = cv2.bitwise_and(rgb, rgb, mask=red_mask)
green = cv2.bitwise_and(rgb, rgb, mask=green_mask)
blue = cv2.bitwise_and(rgb, rgb, mask=blue_mask)

# Display
plt.figure(figsize=(12,4))

plt.subplot(131)
plt.imshow(red)
plt.title("Red Extracted")
plt.axis("off")

plt.subplot(132)
plt.imshow(green)
plt.title("Green Extracted")
plt.axis("off")

plt.subplot(133)
plt.imshow(blue)
plt.title("Blue Extracted")
plt.axis("off")

plt.show()
```
