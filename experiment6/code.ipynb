```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
signal=np.array([1,2,3,4,5,6,7,8])
fft_res=np.fft.fft(signal)
magni_spec=np.abs(fft_res)
phs_spec=np.angle(fft_res)
print(fft_res)
print(magni_spec)
print(magni_spec)
# for 2D-fft2
```

```python
reconstructed_signal =np.fft.ifft(fft_res)

# rnge,signal length
sample_idx=np.arange(len(signal))
```

```python
plt.figure(figsize=(10, 6))

plt.subplot(2,2,1)
plt.stem(sample_idx, signal)
plt.title("Original Spectrum")
plt.xlabel("Sample Index")
plt.ylabel("Amplitude")

plt.subplot(2, 2, 2)
plt.stem(sample_idx, magni_spec)
plt.title("Magnitude Spectrum")
plt.xlabel("Sample Index")
plt.ylabel("Magnitude")

plt.subplot(2, 2, 3)
plt.stem(sample_idx, phs_spec)
plt.title("Phase Spectrum")
plt.xlabel("Sample Index")
plt.ylabel("Phase")

plt.subplot(2, 2, 4)
plt.stem(sample_idx, reconstructed_signal)
plt.title("Reconstructed signal")
plt.xlabel("Sample Index")
plt.ylabel("Amplitude")

plt.tight_layout()
plt.show()
```
