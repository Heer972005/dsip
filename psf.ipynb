```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
img=cv2.imread("image.png")

# astype-type casting
img=img.astype(np.float32)/255.0
```

```python
size=15
center=size//2
```

```python
x=np.arange(-7,8)
X,Y=np.meshgrid(x,x)
sigma=3
gauss_psf=np.exp(-(X**2+Y**2)/(2*sigma**2))
gauss_psf=gauss_psf/np.sum(gauss_psf)
```

```python
# horishontal psf
hori_psf=np.zeros((size,size),dtype=np.float32)
hori_psf[center,:]=1
hori_psf=hori_psf/np.sum(hori_psf)
```

```python
# vertical psf
ver_psf=np.zeros((size,size),dtype=np.float32)
ver_psf[:,center]=1
ver_psf=hori_psf/np.sum(ver_psf)
```

```python
# horishontal psf
diag_psf=np.zeros((size,size),dtype=np.float32)
np.fill_diagonal(diag_psf,1)
hori_psf=hori_psf/np.sum(diag_psf)
```

```python
# horishontal psf
cir_psf=np.zeros((size,size),dtype=np.float32)
Yc,Xc=np.ogrid[:size,:size]
dis=np.sqrt((Xc-center)**2+(Yc-center)**2)
cir_psf[dis<=6]=1
cir_psf=cir_psf/np.sum(cir_psf);
```

```python
ring_psf=np.zeros((size,size),dtype=np.float32)
ring_psf[(dis>=4)&(dis<=6)]=1
ring_psf=ring_psf/np.sum(ring_psf);
```
