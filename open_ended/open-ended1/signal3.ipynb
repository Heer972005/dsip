```python
import numpy as np
import matplotlib.pyplot as plt

n = np.arange(-2,3)

x = []

for i in n:
    if i < 0:
        x.append(-2)
    elif i <= 1:
        x.append(i)
    else:
        x.append(2)

plt.figure(figsize=(6,4))
plt.plot(n,x)
plt.title("Signal 3")
plt.xlabel("n")
plt.ylabel("x[n]")
plt.grid(True)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

n1 = [-2, -1]
x1 = [-2, -2]

n2 = np.linspace(-1, 1, 100)
x2 = 2 * n2

n3 = [1, 2]
x3 = [2, 2]

plt.figure(figsize=(6,4))

plt.plot(n1, x1, linewidth=2)
plt.plot(n2, x2, linewidth=2)
plt.plot(n3, x3, linewidth=2)

# Vertical drop at n = 2
plt.plot([2,2],[2,0], linewidth=2)

plt.title("Signal 3")
plt.xlabel("n")
plt.ylabel("x[n]")

plt.xticks(np.arange(-2,3,1))
plt.yticks(np.arange(-2,3,1))
plt.grid(True)

plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

def line(n):
    return 2 * n

n = np.linspace(-2, 2, 400)

x = np.piecewise(
    n,
    [
        n < -1,
        (n >= -1) & (n <= 1),
        (n > 1) & (n <= 2)
    ],
    [
        -2,
        line,
        2
    ]
)

plt.figure(figsize=(6,4))

plt.plot(n, x, linewidth=2)

# Vertical lines
plt.plot([2, 2], [2, 0], 'b',linewidth=2)
plt.plot([-2, -2], [0, -2], 'b',linewidth=2)

plt.title("Signal 3")
plt.xlabel("n")
plt.ylabel("x[n]")

plt.xticks(np.arange(-2, 3, 1))
plt.yticks(np.arange(-2, 3, 1))

plt.axhline(y=0, color='black', linewidth=1)
plt.grid(True)

plt.show()
```
