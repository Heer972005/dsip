```python
import numpy as np
import matplotlib.pyplot as plt

def linear_convo(s1,s2):
    linear_convo=np.convolve(s1,s2,mode="full")
    return linear_convo

def circular_convo(s1,s2):
    fft_length=max(len(s1),len(s2))
    fft_s1=np.fft.fft(s1,fft_length)
    fft_s2=np.fft.fft(s2,fft_length)

    circular_convo=np.fft.ifft(fft_s1*fft_s2)
    return np.real(circular_convo)


s11=np.array([1,2,3,4,5])
s22=np.array([2,4,6,8,10])

linear_conv = linear_convo(s11, s22)
circular_conv = circular_convo(s11, s22)

plt.figure(figsize=(10, 25))
plt.subplot(2, 1, 1)
plt.stem(linear_conv)
plt.title('Linear Convolution')
plt.xlabel('Sample')
plt.ylabel('Amplitude')
plt.yticks(np.arange(0,100, 1))
plt.subplot(2, 1, 2)
plt.stem(circular_conv)
plt.title('Circular Convolution')
plt.xlabel('Sample')
plt.ylabel('Amplitude')
plt.tight_layout();
plt.show()
```
