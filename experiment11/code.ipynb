```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
image=cv2.imread('image.png',0)
```

```python
kernel=np.ones((5,5),np.uint8)
```

```python
img_dilated=cv2.dilate(image,kernel,iterations = 1)
img_dilated2=cv2.dilate(image,kernel,iterations=2)
```

```python
plt.imshow(image,cmap='gray')
```

```python
# plt.imshow(image,cmap='gray')
plt.imshow(img_dilated,cmap='gray')
```

```python
eroded_img=cv2.erode(img_dilated,kernel,iterations = 1)
plt.imshow(eroded_img,cmap='gray')
```

```python
open_img=cv2.morphologyEx(image,cv2.MORPH_OPEN,kernel,iterations=2)
plt.imshow(open_img,cmap='gray')
```

```python
close_img=cv2.morphologyEx(image,cv2.MORPH_CLOSE,kernel,iterations=2)
plt.imshow(close_img,cmap='gray')
```
